## Task 3: Determine Trimming Parameters

To create the configuration file, you need to figure out what adapters and primers to trim.

### Step 1: Understand adapters and primers

**Question:**

**B3.1)** Why is the adapter the same for exon 7 as for exon 5?


### Step 2: Identify the primers

The primers are **specific to each library** because they're used to amplify the HDR template.

**How to find them:**

For this exercise, we'll give you the primers:

```
Forward primer: GCTGTGGGAGCTGATGTGGGG
Reverse primer: GAGCTGGGGCTCAGGGCCCTCTGGTATGT
```

**Why primers differ between exons:**
* Primers are designed to amplify the specific HDR template  
* Each exon is at a different genomic location  
* Primers must match the flanking genomic sequence  
* This is **biology-specific**, unlike adapters which are **technology-specific**

## Task 5: Create the Configuration File

Now create `quants_7a.json` in the `data/exon7/` directory.

*Note: this library only has 22 columns so `mseq` or the oligo sequences (python library convertor `-S`) is 22 for exon 7.

**Starting template:**

```
{
  "max_cpus": 4,
  "single_end": true,
  "input_type": "fastq",
  "raw_sequencing_qc": true,
  "adapter_trimming": "cutadapt",
  "adapter_trimming_qc": true,
  "adapter_cutadapt_options": "[YOUR_ADAPTER_SEQUENCE]",
  "primer_trimming": "cutadapt",
  "primer_trimming_qc": true,
  "primer_cutadapt_options": "[YOUR_PRIMER_SEQUENCES]",
  "read_modification": true,
  "append_start": "[YOUR_APPEND_START]",
  "append_end": "[YOUR_APPEND_END]",
  "append_quality": "?",
  "transform_library": true,
  "pyquest_library_converter_options": "-N 1 -S 22",
  "quantification": "pyquest"
}
```

Fill in the bracketed values using what you learned in Tasks 3 and 4.

**Key points:**
- Adapter: Same as exon 5 (Illumina infrastructure)
- Primers: Exon 7-specific (linked with `...`)
- Append sequences: Same as exon 5 (synthesis method)
