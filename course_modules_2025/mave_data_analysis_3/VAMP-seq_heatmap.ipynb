# Quick heatmap from the VAMP-seq example

```python
import pandas as pd
import matplotlib.pyplot as plt
```

First we're going to read the data in from the CountESS example.

```python
df = pd.read_csv("G6PD_G6P_Domain_CountESS_scores.csv")
df
```

This code decomposes the `hgvsp` column into the variant amino acid and the position.
Special handling is included for synonymous variants.

```python
df['variant_aa'] = [x[2:5] if x.endswith("=") else x[-3:] for x in df['hgvsp']]
df['variant_position'] = [int(x[5:-1]) if x.endswith("=") else int(x[5:-3]) for x in df['hgvsp']]
```

```python
df[[x.endswith("=") for x in df['hgvsp']]]
```

To make a heatmap in matplotlib, we need to create a matrix representation of the data frame using the `pivot` function.

```python
plot_data = df.pivot(index="variant_aa", columns="variant_position", values="score")
plot_data
```

Now we can generate the heatmap itself and the color bar.
This is a very simple example, and there are many more annotations that could be added, such as marking the wild-type residues.

For more information about creating heatmaps in matplotlib see the [Annotated heatmap documentation](https://matplotlib.org/stable/gallery/images_contours_and_fields/image_annotated_heatmap.html)
or, for an older DMS-specific solution, the [Enrich2 source code](https://github.com/FowlerLab/Enrich2).

```python
fig, ax = plt.subplots()
im = ax.imshow(plot_data, cmap="RdBu_r")
ax.set_xticks(range(len(plot_data.columns)), labels=plot_data.columns, rotation=45)
ax.set_yticks(range(len(plot_data.index)), labels=plot_data.index)
cbar = ax.figure.colorbar(im, ax=ax)
```
